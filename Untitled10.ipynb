```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
train_texts = [
    text for text in ds['train'][:100]['text']
    if text.strip()
]
```

```python
tokenizer.fit_on_texts(train_texts)
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
input_seq = []

for sentence in train_texts :
  tok_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1, len(tok_sentence)):
    input_seq.append(tok_sentence[:i+1])
```

```python
input_seq[0:10]
```

```python
max_len = max([len(x) for x in input_seq])
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_seq = pad_sequences(input_seq, maxlen=max_len, padding='pre')
```

```python
padded_input_seq
```

```python
X = padded_input_seq[:,:-1]
```

```python
y = padded_input_seq[:,-1]
```

```python
X.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical

y = to_categorical(y,num_classes=len(tokenizer.word_index)+1)
```

```python
y
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Input
```

```python
model = Sequential([
    Input(shape=(max_len - 1,)),
    Embedding(1668, 100),
    LSTM(150),
    Dense(1668, activation='softmax')
])
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
print(X.shape, X.dtype)
print(y.shape, y.dtype)
print(max_len)
```

```python
model.fit(X,y,epochs=80)
```

```python
import time
import numpy as np

text = "Bush Theatre in the"

for i in range(10) :

  token_text = tokenizer.texts_to_sequences([text])[0]
  padded_token_text = pad_sequences([token_text], maxlen = max_len, padding='pre')
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items() :
    if index == pos :
      text = text + " " + word
      print(text)
      time.sleep(2)
```
